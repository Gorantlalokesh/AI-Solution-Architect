```python
import numpy as np

scores = np.array([70, 80, 90, 60, 100])

print("Sum:", np.sum(scores))
print("Mean:",np.mean(scores))
print("Median",np.median(scores))
print("Minimum",np.min(scores))
print("Maximum",np.max(scores))
print("Standard deviation",np.std(scores))
```

```python
import numpy as np

X = np.array([[1, 2],
              [3, 4]])

Y = np.array([[5, 6],
              [7, 8]])

print("\nVertical stack:\n", np.vstack([X, Y]))

print("\nHorizontal stack:\n", np.hstack([X, Y]))
```

```python
import numpy as np

rng = np.random.default_rng(42)

print("Random integers:", rng.integers(1, 10, size=8))
print("Random floats:", rng.random(5))
print("Normal distribution:", rng.normal(size=5))
```

```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print("Element-wise multiplication:\n", A * B)

print("\nMatrix multiplication:\n", A @ B)

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("\nDot product:", np.dot(a, b))
```

```python
import numpy as np

data = np.array([10, 20, np.nan, 40, 50])

print("Is NaN:", np.isnan(data))
print("Normal mean:", np.mean(data))
print("NaN-safe mean:", np.nanmean(data))
print("NaN-safe sum:", np.nansum(data))
```
